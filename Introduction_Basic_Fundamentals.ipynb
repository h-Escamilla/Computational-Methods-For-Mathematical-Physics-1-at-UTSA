```python
#summing over integers 1 to 4
n = 0
#So in python, indentation matters. Everything inside the loop is indented. To code outside we need to leave the indentation
for i in range(1, 5):
    n = n + i
n_sum = n
print(n_sum)
```

## Plotting

```python
n = 15 #here we are declaring the number of points we will use
x = np.linspace(0, 10, n+1)   # We are creating a number of equally distanced points n+1 between x = 0 and x = 10.
y = np.sin(x)
print(y)
plt.plot(x,y)   #plt calls the plot function from matplotlib because we defined that way
```

This Graph looks bad. We don't have enough points to generate a nice looking sin(x). The solution is to increase n.

```python
n = 100
x = np.linspace(0, 10, n+1)
y = np.sin(x)
print(y)  #this just prints the y values. since we are using n+1 we are generating that many y values.
plt.plot(x, y)
```

# Plotting Example 1
Plot Sin(x) and Cos(x) on the same graph.
Inlcude Title, label your axis, legend, differentiate the trig functions with different colors


```python
n = 1000
x = np.linspace(0, 10, n+1)
y1 = np.sin(x)
y2 = np.cos(x)

#Plotting
#You can look up different ways to customize your graphs
#Most of this is personal preference
plt.figure(figsize=(8,3)) #Determines the size of our image
plt.plot(x, y1, label='Sine Wave', color='blue', linewidth =2)
plt.plot(x, y2, label='Cosine Wave', color='red', linestyle='--', linewidth=2)
plt.title('Sin(x) and Cos(x)', fontsize=14, fontweight='bold')
plt.xlabel("Time (seconds)", fontsize=11)
plt.ylabel("Amplitude", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6, color='gray')
plt.legend(loc='best', fontsize=7, shadow=True)
plt.show() 
```

# Problem 1

The equation for the height of a thrown ball is $y = -\frac{1}{2}gt^2 + v_0t$

Jon throws his ball at 14 m/s and Rob throws his at 10 m/s upward. Plot the height of each ball as a function of time from 0 to 2 seconds. Who threw the ball the highest

Finish the partially written code and explain what each line is doing


```python
#Declare all the variables you need
n = 1000
t = np.linspace(0, 2, n+1) #seconds
g = 9.81 #m/s
v0_j = 14 #m/s
v0_r = 10 #m/s

#Write the equations 
y_j = -1/2 * g * t**2 + v0_j*t
y_r = -1/2 * g * t**2 + v0_r*t

#Plot them
plt.plot(t, y_j, label='Jon')
plt.plot(t, y_r, label='Rob')
plt.xlabel('Time [seconds]')
plt.ylabel('Height [meters]')
plt.title('Jon vs Rob')
plt.grid(True)
plt.legend()
plt.show()
```

# Problem 2
Plot sin(x) shifted by 90 degrees to the left against cos(x)

Note: Numpy trig functions take in pi radians. You need to convert degrees to radians for the numpy function to work

```python
#declaring variables
n = 100
x = np.linspace(0, 10, n+1)
y1 = np.sin(x + np.pi/2)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x + 90)', linewidth=3)
plt.plot(x, y2, label='cos(x)', linestyle ='--', color='red')
plt.xlabel('Time [seconds]')
plt.ylabel('Amplitude [meters]')
plt.title('Sin(x + 90) and Cos(x)')
plt.grid(True)
plt.legend()
plt.show()
```
